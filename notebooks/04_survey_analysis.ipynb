## 2. 메시지·그룹별 평균

```python
groups = {
    "원래": ["앞으로도 좋은 성과", "숙제를 훌륭하게 마쳤고", "후회 없이 수고한 숙제"],
    "학습 이전": ["아이유 챌린지 7회차", "IVE 멋진 결과", "에스파 챌린지 성공"],
    "학습 이후": ["에스파 다양한 매력", "와우 숙제 완료", "IVE ATTITUDE"]
}

# 메시지별 평균/표준편차 계산
msg_stats = {}
for msg, scores in data.items():
    arr = np.array(scores, dtype=float)
    msg_stats[msg] = {
        "평균": np.nanmean(arr),
        "표준편차": np.nanstd(arr, ddof=1)
    }

df_msg = pd.DataFrame(msg_stats).T

# 메시지별 평균 내림차순 정렬
df_msg_sorted = df_msg.sort_values("평균", ascending=False)

# 그룹별 평균 계산
group_stats = {}
for g, msgs in groups.items():
    all_scores = []
    for msg in msgs:
        all_scores.extend(data[msg])
    arr = np.array(all_scores, dtype=float)
    group_stats[g] = np.nanmean(arr)

df_group = pd.DataFrame.from_dict(group_stats, orient="index", columns=["그룹 평균"])
df_group_sorted = df_group.sort_values("그룹 평균", ascending=False)

# 출력
print("메시지별 평균/표준편차 (내림차순):")
print(df_msg_sorted, "\n")
print("그룹별 평균 (내림차순):")
print(df_group_sorted)
```

```python
import numpy as np

# 메시지 × 응답자 행렬로 변환 (9행 × 17열)
matrix = np.array(list(data.values()), dtype=float)

# 동일 응답자 판별
cols_to_keep = []
for j in range(matrix.shape[1]):  # 각 응답자(열) 검사
    col = matrix[:, j]
    col_nonan = col[~np.isnan(col)]
    if len(set(col_nonan)) > 1:   # 값이 2개 이상이면 keep
        cols_to_keep.append(j)

# 필터링 적용
filtered = {msg: [scores[j] for j in cols_to_keep] for msg, scores in data.items()}

print("삭제된 응답자 수:", matrix.shape[1] - len(cols_to_keep))
print("남은 응답자 수:", len(cols_to_keep))
print(filtered)

reliability_data = list(filtered.values())

# Krippendorff’s alpha (ordinal scale)
alpha = krippendorff.alpha(reliability_data=reliability_data, level_of_measurement='ordinal')
print("Krippendorff’s Alpha (ordinal):", alpha)
```

## 3. 숙제 완료율 예측과 메시지 점수의 관계

```python
import numpy as np
from scipy.stats import spearmanr

# -----------------------
# 데이터 (6개 메시지)
# -----------------------
responses = {
    "에스파 다양한 매력": [2,2,np.nan,3,1,3,4,4,4,4,3,3,2,5,4,5,1],
    "와우 숙제 완료":     [5,5,5,3,1,3,4,5,5,5,3,3,1,5,4,5,5],
    "아이유 챌린지 7회차": [5,3,np.nan,4,1,3,4,5,4,4,3,3,1,5,4,5,5],
    "IVE 멋진 결과":      [5,4,5,2,1,3,4,3,4,4,3,3,1,5,4,5,3],
    "IVE ATTITUDE":       [1,1,5,2,1,3,4,3,3,4,3,3,1,5,4,5,3],
    "에스파 챌린지 성공": [5,5,5,3,1,3,4,5,4,4,4,3,1,5,4,5,5]
}

labels = [0, 1, 1, 0, 0, 1]   # 정답 라벨 (x=0, o=1)

# -----------------------
# 응답자 기준으로 배열 만들기
# -----------------------
msg_names = list(responses.keys())
arr = np.array([responses[m] for m in msg_names], dtype=float)  # (6 × 응답자수)
arr = arr.T  # (응답자수 × 6)

# -----------------------
# 동일 응답자 제거
# -----------------------
mask_valid = []
for row in arr:
    vals = row[~np.isnan(row)]
    mask_valid.append(not (len(np.unique(vals)) == 1))  # 모두 같은 값이면 제거

mask_valid = np.array(mask_valid)
arr_clean = arr[mask_valid]

print("원래 응답자 수:", arr.shape[0])
print("동일 응답자 제거 후:", arr_clean.shape[0])

# -----------------------
# 메시지별 다시 dict로 변환
# -----------------------
responses_clean = {msg_names[i]: arr_clean[:, i].tolist() for i in range(len(msg_names))}

# -----------------------
# 메시지별 평균 (NA 제외)
# -----------------------
msg_means = {m: np.nanmean(responses_clean[m]) for m in msg_names}

print("\n메시지별 평균 점수 (동일응답자 제거 후)")
for m, v in msg_means.items():
    print(f"  {m}: {v:.3f}")
```

```python
import numpy as np
from scipy.stats import spearmanr

# -----------------------
# 데이터 (6개 메시지)
# -----------------------
responses = {
    "에스파 다양한 매력": [2,2,np.nan,3,1,3,4,4,4,4,3,3,2,5,4,5,1],
    "와우 숙제 완료":     [5,5,5,3,1,3,4,5,5,5,3,3,1,5,4,5,5],
    "아이유 챌린지 7회차": [5,3,np.nan,4,1,3,4,5,4,4,3,3,1,5,4,5,5],
    "IVE 멋진 결과":      [5,4,5,2,1,3,4,3,4,4,3,3,1,5,4,5,3],
    "IVE ATTITUDE":       [1,1,5,2,1,3,4,3,3,4,3,3,1,5,4,5,3],
    "에스파 챌린지 성공": [5,5,5,3,1,3,4,5,4,4,4,3,1,5,4,5,5]
}

labels = [0, 1, 1, 0, 0, 1]   # 정답 라벨 (x=0, o=1)

msg_names = list(responses.keys())

# -----------------------
# 응답자 기준으로 전치 (응답자 × 메시지)
# -----------------------
arr = np.array([responses[m] for m in msg_names], dtype=float).T

# -----------------------
# 동일 응답자 제거
# -----------------------
mask_valid = []
for row in arr:
    vals = row[~np.isnan(row)]
    mask_valid.append(not (len(np.unique(vals)) == 1))  # 모두 동일하면 제거
mask_valid = np.array(mask_valid)
arr_clean = arr[mask_valid]

print("원래 응답자 수:", arr.shape[0])
print("동일 응답자 제거 후:", arr_clean.shape[0])

# -----------------------
# 메시지별 다시 dict 변환
# -----------------------
responses_clean = {msg_names[i]: arr_clean[:, i].tolist() for i in range(len(msg_names))}

# -----------------------
# 메시지별 평균
# -----------------------
msg_means = {m: np.nanmean(responses_clean[m]) for m in msg_names}
means_vector = np.array([msg_means[m] for m in msg_names])

print(msg_means)
print(means_vector)
# -----------------------
# 1) Spearman 상관 (메시지 평균 ↔ 라벨)
# -----------------------
rho, pval = spearmanr(means_vector, labels)
print("\nSpearman 상관")
print(f"  rho={rho:.3f}, p={pval:.4f}")

# -----------------------
# 평가 지표 함수
# -----------------------
def bin_metrics(y_true, y_pred):
    y_true = np.array(y_true, dtype=int)
    y_pred = np.array(y_pred, dtype=int)
    tp = int(((y_true==1)&(y_pred==1)).sum())
    tn = int(((y_true==0)&(y_pred==0)).sum())
    fp = int(((y_true==0)&(y_pred==1)).sum())
    fn = int(((y_true==1)&(y_pred==0)).sum())
    acc = (tp+tn)/(tp+tn+fp+fn)
    prec = tp/(tp+fp) if (tp+fp)>0 else np.nan
    rec = tp/(tp+fn) if (tp+fn)>0 else np.nan
    f1 = 2*prec*rec/(prec+rec) if (prec is not np.nan and rec is not np.nan and (prec+rec)>0) else np.nan
    return {"acc":acc, "precision":prec, "recall":rec, "f1":f1, "tp":tp, "tn":tn, "fp":fp, "fn":fn}

THRESH = 4.0

# -----------------------
# 2) 메시지 단위 분류 (평균 점수 ≥4 → 완료 예측)
# -----------------------
y_pred_msg = (means_vector >= THRESH).astype(int)
mtr_msg = bin_metrics(labels, y_pred_msg)

print(f"\n메시지 단위 분류 (임계값 {THRESH})")
print(mtr_msg)

# -----------------------
# 3) 개별 응답 단위 분류 (점수 ≥4 → 완료 예측)
# -----------------------
y_true_indiv, y_pred_indiv = [], []
for i, m in enumerate(msg_names):
    arr = np.array(responses_clean[m], dtype=float)
    mask = ~np.isnan(arr)
    y_true_indiv.extend([labels[i]] * int(mask.sum()))
    y_pred_indiv.extend((arr[mask] >= THRESH).astype(int).tolist())

mtr_ind = bin_metrics(y_true_indiv, y_pred_indiv)

print(f"\n개별 응답 단위 분류 (임계값 {THRESH})")
print(mtr_ind)
```
